```python
import ga_task2
import utils
import time
import numpy as np
import pygad
import init_pop_greedy
import copy
import matplotlib.pyplot as plt
```

```python
classes = utils.parse_classes("../sem1_files/Classes.txt")
rooms = utils.parse_rooms("../sem1_files/Classrooms.txt")
teachers = utils.parse_teachers("../sem1_files/Teachers.txt")

problem = utils.Problem(classes, rooms, teachers)
```

```python
results={}
utils.reset_metrics()
ga=ga_task2.make_run_ga_task1(problem)
results["normal"]=copy.deepcopy(utils.metrics)
results["normal"]["final_fitness"]=ga.best_solution()[1]
print("Normal", "| fitness:", results["normal"]["final_fitness"],
          "| generacij:", len(utils.metrics["best_fitness"]),
          "| čas: %.2fs" % utils.metrics["gen_time"][-1])
```

```python

utils.reset_metrics()
ga=ga_task2.make_run_ga_task4(problem)
results["init_custom"]=copy.deepcopy(utils.metrics)
results["init_custom"]["final_fitness"]=ga.best_solution()[1]
print("init_custom", "| fitness:", results["init_custom"]["final_fitness"],
          "| generacij:", len(utils.metrics["best_fitness"]),
          "| čas: %.2fs" % utils.metrics["gen_time"][-1])
```

```python
steps = ["fitness", "parents", "crossover", "mutation"]

for name, res in results.items():
    fig, ax = plt.subplots(1, 3, figsize=(16, 4))
    fig.suptitle(name)

    # 1) fitness po generacijah
    ax[0].plot(res["best_fitness"])
    ax[0].set_xlabel("generacija")
    ax[0].set_ylabel("najboljši fitness")
    ax[0].set_title("Fitness po generacijah")

    # 2) fitness po realnem času
    ax[1].plot(res["gen_time"], res["best_fitness"])
    ax[1].set_xlabel("čas [s]")
    ax[1].set_ylabel("najboljši fitness")
    ax[1].set_title("Fitness po času")

    # 3) čas po korakih (naloženi stolpec)
    ax[2].bar(steps, [res["times"][s] for s in steps])
    ax[2].set_ylabel("čas [s]")
    ax[2].set_title("Čas po korakih")

    plt.tight_layout()
    plt.show()
```
